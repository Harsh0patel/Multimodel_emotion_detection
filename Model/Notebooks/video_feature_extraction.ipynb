```python
import os
import cv2
```

```python
video_file = "C:/Users/hp333/Desktop/Multimodel_emotion_detection/data/MELD.Raw/train/train_splits/dia0_utt0.mp4"
output_dir = "output"

cap = cv2.VideoCapture(video_file)
frame_idx = 0
saved_count = 0

os.makedirs(output_dir, exist_ok=True)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    output_path = os.path.join(output_dir, f"frame_{frame_idx:05d}.jpg")
    cv2.imwrite(output_path, frame)
    saved_count += 1
    
    frame_idx += 1

cap.release()
```
